```python
# Import required libraries

import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
from datetime import datetime

print("Libraries imported successfully!")
```

```python
# Connect to SQLite database
conn = sqlite3.connect("finance.db")

cursor = conn.cursor()

print("Database connected successfully!")
```

```python
# Create Income table
cursor.execute("""
CREATE TABLE IF NOT EXISTS income (
    income_id INTEGER PRIMARY KEY AUTOINCREMENT,
    date TEXT NOT NULL,
    source TEXT NOT NULL,
    amount REAL NOT NULL
)
""")

# Create Expenses table
cursor.execute("""
CREATE TABLE IF NOT EXISTS expenses (
    expense_id INTEGER PRIMARY KEY AUTOINCREMENT,
    date TEXT NOT NULL,
    category TEXT NOT NULL,
    description TEXT,
    amount REAL NOT NULL,
    payment_method TEXT
)
""")

# Create Budget table
cursor.execute("""
CREATE TABLE IF NOT EXISTS budgets (
    budget_id INTEGER PRIMARY KEY AUTOINCREMENT,
    month TEXT NOT NULL,
    category TEXT NOT NULL,
    budget_amount REAL NOT NULL
)
""")

# Save changes
conn.commit()

print("All tables created successfully!")
```

```python
# Sample income data

income_data = [
    ("2026-09-01", "Pocket Money", 5000),
    ("2026-09-05", "Part-time Work", 3000),
    ("2026-09-10", "Scholarship", 2000)
]

cursor.executemany("""
INSERT INTO income (date, source, amount)
VALUES (?, ?, ?)
""", income_data)

conn.commit()

print("Sample income added successfully!")
```

```python
# Sample expense data

expense_data = [
    ("2026-09-01", "Food", "Lunch", 120, "UPI"),
    ("2026-09-02", "Transport", "Bus", 50, "Cash"),
    ("2026-09-03", "Education", "Notebook", 200, "UPI"),
    ("2026-09-04", "Food", "Snacks", 80, "UPI"),
    ("2026-09-05", "Shopping", "Clothes", 800, "Card"),
    ("2026-09-06", "Transport", "Auto", 150, "UPI"),
    ("2026-09-07", "Food", "Lunch", 150, "UPI"),
    ("2026-09-08", "Entertainment", "Movie", 250, "UPI"),
    ("2026-09-09", "Education", "Books", 500, "UPI"),
    ("2026-09-10", "Food", "Dinner", 200, "Cash")
]

cursor.executemany("""
INSERT INTO expenses
(date, category, description, amount, payment_method)
VALUES (?, ?, ?, ?, ?)
""", expense_data)

conn.commit()

print("Sample expenses added successfully!")
```

```python
income_df = pd.read_sql_query(
    "SELECT * FROM income",
    conn
)

income_df
```

```python
expense_df = pd.read_sql_query(
    "SELECT * FROM expenses",
    conn
)

expense_df
```

```python
total_income = income_df["amount"].sum()

print(f"Total Income: ₹{total_income:,.2f}")
```

```python
total_expenses = expense_df["amount"].sum()

print(f"Total Expenses: ₹{total_expenses:,.2f}")
```

```python
savings = total_income - total_expenses

print(f"Total Income : ₹{total_income:,.2f}")
print(f"Total Expense: ₹{total_expenses:,.2f}")
print(f"Savings      : ₹{savings:,.2f}")
```

```python
if total_income > 0:
    savings_percentage = (savings / total_income) * 100
else:
    savings_percentage = 0

print(f"Savings Percentage: {savings_percentage:.2f}%")
```

```python
category_expenses = (
    expense_df
    .groupby("category")["amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Expense by Category:")
print(category_expenses)
```

```python
fig = px.pie(
    values=category_expenses.values,
    names=category_expenses.index,
    title="Expenses by Category"
)

fig.show()
```

```python
fig = px.bar(
    x=category_expenses.index,
    y=category_expenses.values,
    title="Category-wise Expense",
    labels={
        "x": "Expense Category",
        "y": "Amount (₹)"
    }
)

fig.show()
```

```python
summary = pd.DataFrame({
    "Financial Metric": [
        "Total Income",
        "Total Expenses",
        "Savings",
        "Savings Percentage"
    ],
    "Value": [
        total_income,
        total_expenses,
        savings,
        f"{savings_percentage:.2f}%"
    ]
})

summary
```

```python
budget_data = [
    ("2026-09", "Food", 1500),
    ("2026-09", "Transport", 1000),
    ("2026-09", "Education", 1000),
    ("2026-09", "Shopping", 1000),
    ("2026-09", "Entertainment", 500)
]

cursor.executemany("""
INSERT INTO budgets
(month, category, budget_amount)
VALUES (?, ?, ?)
""", budget_data)

conn.commit()

print("Budget data added successfully!")
```

```python
budget_df = pd.read_sql_query(
    "SELECT * FROM budgets",
    conn
)

actual = (
    expense_df
    .groupby("category")["amount"]
    .sum()
    .reset_index()
)

budget_analysis = budget_df.merge(
    actual,
    on="category",
    how="left"
)

budget_analysis.rename(
    columns={"amount": "actual_expense"},
    inplace=True
)

budget_analysis["actual_expense"] = (
    budget_analysis["actual_expense"].fillna(0)
)

budget_analysis["variance"] = (
    budget_analysis["actual_expense"]
    - budget_analysis["budget_amount"]
)

budget_analysis["status"] = np.where(
    budget_analysis["variance"] > 0,
    "Over Budget",
    "Within Budget"
)

budget_analysis
```
